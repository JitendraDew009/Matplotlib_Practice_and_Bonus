```python
#ploting two graphs over here + adding labels + lagend (important to to show the label)
x = [1, 2, 3, 4]
dsa_marks = [90, 92, 96, 98] #blue
ml_marks = [93, 95, 98, 100] #orange
plt.plot(x,dsa_marks, label = 'dsa_marks')  #label added
plt.plot(x,ml_marks, label = 'ml_marks') #label added
plt.legend(loc = "lower right") # lower side showcase **IMPORTANT**

# plt.plot(x,y) # X= 1,2,3,4 and Y axis = 4, 5, 6, 7 
plt.title("Marks corresponding to no. of hours")
plt.xlabel("No. of hrs studied")
plt.ylabel("Marks")
plt.xticks([1,2,4]) #show only numbers as you want
plt.yticks([90,95,100]) #show only numbers as you want
plt.show()
```

```python
#ploting two graphs over here + adding labels + color added
x = [1, 2, 3, 4]
dsa_marks = [90, 92, 96, 98] #blue
ml_marks = [93, 95, 98, 100] #orange
plt.plot(x,dsa_marks, label = 'dsa_marks', color = 'red')  #label added
plt.plot(x,ml_marks, label = 'ml_marks', color = 'blue') #label added
plt.legend(loc = "lower right") # lower side showcase  



# plt.plot(x,y) # X= 1,2,3,4 and Y axis = 4, 5, 6, 7 
plt.title("Marks corresponding to no. of hours")
plt.xlabel("No. of hrs studied")
plt.ylabel("Marks")
plt.xticks([1,2,4]) #show only numbers as you want
plt.yticks([90,95,100]) #show only numbers as you want
plt.show()
```

```python
#ploting two graphs over here + adding labels + color added + linewidth added
x = [1, 2, 3, 4]
dsa_marks = [90, 92, 96, 98] #blue
ml_marks = [93, 95, 98, 100] #orange
plt.plot(x,dsa_marks, label = 'dsa_marks', color = 'red', linewidth = 5)  #label added
plt.plot(x,ml_marks, label = 'ml_marks', color = 'blue', linewidth = 6) #label added
plt.legend(loc = "lower right") # lower side showcase  



# plt.plot(x,y) # X= 1,2,3,4 and Y axis = 4, 5, 6, 7 
plt.title("Marks corresponding to no. of hours")
plt.xlabel("No. of hrs studied")
plt.ylabel("Marks")
plt.xticks([1,2,4]) #show only numbers as you want
plt.yticks([90,95,100]) #show only numbers as you want
plt.show()
```

```python
#ploting two graphs over here + adding labels + color added + linewidth added and changed value + markr added
x = [1, 2, 3, 4]
dsa_marks = [90, 92, 96, 98] #blue
ml_marks = [93, 95, 98, 100] #orange
plt.plot(x,dsa_marks, label = 'dsa_marks', color = 'red', linewidth = 1, marker = '>')  #label added
plt.plot(x,ml_marks, label = 'ml_marks', color = 'blue', linewidth = 1.5, marker = '*') #label added
plt.legend(loc = "lower right") # lower side showcase  



# plt.plot(x,y) # X= 1,2,3,4 and Y axis = 4, 5, 6, 7 
plt.title("Marks corresponding to no. of hours")
plt.xlabel("No. of hrs studied")
plt.ylabel("Marks")
plt.xticks([1,2,4]) #show only numbers as you want
plt.yticks([90,95,100]) #show only numbers as you want
plt.show()
```

```python
#ploting two graphs over here + adding labels + color added + linewidth added and changed value + markr added + linestyles added

x = [1, 2, 3, 4]
dsa_marks = [90, 92, 96, 98] #blue
ml_marks = [93, 95, 98, 100] #orange
plt.plot(x,dsa_marks, label = 'dsa_marks', color = 'red', linewidth = 1, marker = '>', linestyle = '--')  #label added
plt.plot(x,ml_marks, label = 'ml_marks', color = 'blue', linewidth = 1.5, marker = '*') #label added
plt.legend(loc = "lower right") # lower side showcase  



# plt.plot(x,y) # X= 1,2,3,4 and Y axis = 4, 5, 6, 7 
plt.title("Marks corresponding to no. of hours")
plt.xlabel("No. of hrs studied")
plt.ylabel("Marks")
plt.xticks([1,2,4]) #show only numbers as you want
plt.yticks([90,95,100]) #show only numbers as you want
plt.show()
```

```python
#ploting two graphs over here + adding labels + color added + linewidth added and changed value + markr added + linestyles added
# code shorten

x = [1, 2, 3, 4]
dsa_marks = [90, 92, 96, 98] #blue
ml_marks = [93, 95, 98, 100] #orange
plt.plot(x,dsa_marks, 'r>--', label = 'dsa_marks')  #code shorted by using same parameter in quote r = red, > = marker and -- for linestyle 
# ----------------------^^^^ this ( 'r>--' ) part should be in alway in between of attributes then it works otherwise not.

plt.plot(x,ml_marks, label = 'ml_marks', color = 'blue', linewidth = 1.5, marker = '*') #label added
plt.legend(loc = "lower right") # lower side showcase  



# plt.plot(x,y) # X= 1,2,3,4 and Y axis = 4, 5, 6, 7 
plt.title("Marks corresponding to no. of hours")
plt.xlabel("No. of hrs studied")
plt.ylabel("Marks")
plt.xticks([1,2,4]) #show only numbers as you want
plt.yticks([90,95,100]) #show only numbers as you want
plt.show()
```

```python
#ploting two graphs over here + adding labels + color added + linewidth added and changed value + markr added + linestyles added
# code shorten + markersize added 

x = [1, 2, 3, 4]
dsa_marks = [90, 92, 96, 98] #blue
ml_marks = [93, 95, 98, 100] #orange
plt.plot(x,dsa_marks, 'r>--', label = 'dsa_marks', markersize = 20  )  #code shorted by using same parameter in quote r = red, > = marker and -- for linestyle 
plt.plot(x,ml_marks, label = 'ml_marks', color = 'blue', linewidth = 1.5, marker = '*') #label added
plt.legend(loc = "lower right") # lower side showcase  



# plt.plot(x,y) # X= 1,2,3,4 and Y axis = 4, 5, 6, 7 
plt.title("Marks corresponding to no. of hours")
plt.xlabel("No. of hrs studied")
plt.ylabel("Marks")
plt.xticks([1,2,4]) #show only numbers as you want
plt.yticks([90,95,100]) #show only numbers as you want
plt.show()
```

```python
#ploting two graphs over here + adding labels + color added + linewidth added and changed value + markr added + linestyles added
# code shorten + markersize added and changed + fontdict added

x = [1, 2, 3, 4]
dsa_marks = [90, 92, 96, 98] #blue
ml_marks = [93, 95, 98, 100] #orange
plt.plot(x,dsa_marks, 'r>--', label = 'dsa_marks', markersize = 10  )  #code shorted by using same parameter in quote r = red, > = marker and -- for linestyle 
plt.plot(x,ml_marks, label = 'ml_marks', color = 'blue', linewidth = 1.5, marker = '*') #label added
plt.legend(loc = "lower right") # lower side showcase  



# plt.plot(x,y) # X= 1,2,3,4 and Y axis = 4, 5, 6, 7 
plt.title("Marks corresponding to no. of hours", fontdict = {'fontname' : 'comic sans ms'} )
plt.xlabel("No. of hrs studied")
plt.ylabel("Marks")
plt.xticks([1,2,4]) #show only numbers as you want
plt.yticks([90,95,100]) #show only numbers as you want
plt.show()
```

```python
#ploting two graphs over here + adding labels + color added + linewidth added and changed value + markr added + linestyles added
# code shorten + markersize added and changed + fontdict added with fontweight : 10 and color : blue

x = [1, 2, 3, 4]
dsa_marks = [90, 92, 96, 98] #blue
ml_marks = [93, 95, 98, 100] #orange
plt.plot(x,dsa_marks, 'r>--', label = 'dsa_marks', markersize = 10  )  #code shorted by using same parameter in quote r = red, > = marker and -- for linestyle 
plt.plot(x,ml_marks, label = 'ml_marks', color = 'blue', linewidth = 1.5, marker = '*') #label added
plt.legend(loc = "lower right") # lower side showcase  



# plt.plot(x,y) # X= 1,2,3,4 and Y axis = 4, 5, 6, 7 
plt.title("Marks corresponding to no. of hours", fontdict = {'fontname' : 'comic sans ms', 'fontweight' : 30, 'color' : 'blue'},  )
plt.xlabel("No. of hrs studied")
plt.ylabel("Marks")
plt.xticks([1,2,4]) #show only numbers as you want
plt.yticks([90,95,100]) #show only numbers as you want
plt.show()
```

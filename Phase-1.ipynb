```python
plt.plot(t, t, 'r--', t, t**2, 'bs', t, t**3) #bs = blue sqaue marker line
```

```python
plt.plot(t, t, 'r--', t, t**2, 'bs', t, t**3, 'g^') #g^ = green triangle marker line
```

```python
plt.plot(t, t, 'r--', t, t**2, 'bs', t, t**3, 'g^') #g^ = green triangle marker line
plt.show()


# x = t
# y = x = t
# (x,y) = (t,t)
#r-- = short hand notation red incolor and -- marker
# //////////////////////////////////////////////////////////////
# x1 = t
# y2 = t**2
# (x1, y1) = (t, t**2)
# bs = show it in blue  with marker of square
# //////////////////////////////////////////////////////////////
# x2 = t
# y2 = t**3
# (x2, y2) = (t, t**3)
# 'g^' = show it in green color and ^ marker.
```

```python
# Plot in different figures
import numpy as np

t = np.arange(0., 5.0, .2)

plt.figure()
plt.plot(t, t, 'r--')
plt.show()

plt.figure()
plt.plot(t, t**2, 'bs')
plt.show()

plt.figure()
plt.plot(t, t**3, 'g^')
plt.show()
```

```python
# Subplots in a figure: one row three columns and 3 subplots horizonatally on each other

t = np.arange(0., 5.0, .2)

plt.figure()

plt.subplot(1, 3, 1)
plt.plot(t, t, 'r--')

plt.subplot(1, 3, 2)
plt.plot(t, t**2, 'bs')

plt.subplot(1, 3, 3)
plt.plot(t, t**3, 'g^')
plt.show()
```

```python
# Subplots in a figure: one row three columns and 3 subplots verticcally on each other

t = np.arange(0., 5.0, .2)

plt.figure()

plt.subplot(3, 1, 1) # 3 row , 1 colum and 1 index
plt.plot(t, t, 'r--')

plt.subplot(3, 1, 2) # 3 row , 1 colum and 2 index
plt.plot(t, t**2, 'bs')

plt.subplot(3, 1, 3) # 3 row , 1 colum and 3 index
plt.plot(t, t**3, 'g^')
plt.show()
```

```python
# plt.suptitle('Different function of t')

t = np.arange(0., 5.0, .2)

plt.figure()

plt.subplot(1, 3, 1)
plt.plot(t, t, 'r--')
plt.title('Linear Subplot')

plt.subplot(1, 3, 2)
plt.plot(t, t**2, 'bs')
plt.title('Quadratic Subplot')

plt.subplot(1, 3, 3)
plt.plot(t, t**3, 'g^')
plt.title(' Cubic Subplot')

plt.suptitle('Different function of t')


plt.show()
```

```python
# plt.tight_layout() added

t = np.arange(0., 5.0, .2)

plt.figure()

plt.subplot(1, 3, 1)
plt.plot(t, t, 'r--')
plt.title('Linear Subplot')

plt.subplot(1, 3, 2)
plt.plot(t, t**2, 'bs')
plt.title('Quadratic Subplot')

plt.subplot(1, 3, 3)
plt.plot(t, t**3, 'g^')
plt.title(' Cubic Subplot')

plt.suptitle('Different function of t')
plt.tight_layout() # layout affect

plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

# Generate 100 points between 1 and 10
x = np.linspace(1, 10, 100)

# Define cotangent and cosecant
def cot(x):
    return 1 / np.tan(x)

def cosec(x):
    return 1 / np.sin(x)

fig, axes = plt.subplots(2, 3, figsize=(12, 8))  # 2 rows, 3 columns

# Sin
axes[0,0].plot(x, np.sin(x))
axes[0,0].set_title('Sin Function')
axes[0,0].set_xlabel('x')
axes[0,0].set_ylabel('y')

# Cos
axes[0,1].plot(x, np.cos(x))
axes[0,1].set_title('Cosine Function')
axes[0,1].set_xlabel('x')
axes[0,1].set_ylabel('y')

# Tan
axes[0,2].plot(x, np.tan(x))
axes[0,2].set_title('Tan Function')
axes[0,2].set_xlabel('x')
axes[0,2].set_ylabel('y')

# Cot
axes[1,0].plot(x, cot(x))
axes[1,0].set_title('Cot Function')
axes[1,0].set_xlabel('x')
axes[1,0].set_ylabel('y')

# Reuse tan (or you could plot sec here instead)
axes[1,1].plot(x, np.tan(x))
axes[1,1].set_title('Tan Function (again)')
axes[1,1].set_xlabel('x')
axes[1,1].set_ylabel('y')

# Cosec
axes[1,2].plot(x, cosec(x))
axes[1,2].set_title('Cosec Function')
axes[1,2].set_xlabel('x')
axes[1,2].set_ylabel('y')

plt.tight_layout()
plt.show()
```

```python
# 3D plot Structure
ax = plt.axes(projection = '3d')

X = np.arange(-5, 5, 0.25)
Y = np.arange(-5, 5, 0.25)

X, Y = np.meshgrid(X,Y) 
R = np.sqrt(X**2 + Y**2) # sqrt = square root 

Z = np.sin(R) # basicall R = (x**2 + y**2)*1/2
ax.plot_surface(X, Y, Z, cmap = 'plasma') #cmap = color change.
plt.show()
```
